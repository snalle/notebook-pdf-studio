# Notebook PDF Studio PoC (Figure Test)

This notebook tests whether Notebook PDF Studio can read and render a static figure stored in a Jupyter notebook output.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y, marker="o")
plt.title("PoC Figure Test")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
